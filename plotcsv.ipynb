```python
import pandas as pd
import matplotlib.pyplot as plt

log = pd.read_csv("training_log_resnet50.csv")
plt.plot(log[log.phase=="train"].epoch, log[log.phase=="train"].loss, label="training loss")
plt.plot(log[log.phase=="val"].epoch, log[log.phase=="val"].loss, label="validation loss")
plt.plot(log[log.phase=="train"].epoch, log[log.phase=="train"].accuracy, label="training accuracy")
plt.plot(log[log.phase=="val"].epoch, log[log.phase=="val"].accuracy, label="validation accuracy")
plt.ylim(0,1)
plt.legend(); plt.show()
```
