```python
import pandas as pd
from matplotlib.ticker import MultipleLocator
import matplotlib.pyplot as plt

log = pd.read_csv("training_log_resnet50.csv")
plt.figure(figsize=(10,8))
plt.gca().yaxis.set_major_locator(MultipleLocator(0.1))
ax = plt.gca()
for e in sorted(log.epoch.unique()):
    ax.axvline(e, color='gray', linestyle='solid', linewidth=0.6, alpha=0.1)
    for y in ax.yaxis.get_majorticklocs():
        ax.axhline(y, color='gray', linestyle='solid', linewidth=0.6, alpha=0.005)
plt.plot(log[log.phase=="train"].epoch, log[log.phase=="train"].loss, label="training loss")
plt.plot(log[log.phase=="val"].epoch, log[log.phase=="val"].loss, label="validation loss")
plt.plot(log[log.phase=="train"].epoch, log[log.phase=="train"].accuracy, label="training accuracy")
plt.plot(log[log.phase=="val"].epoch, log[log.phase=="val"].accuracy, label="validation accuracy")
plt.ylim(0,1)
plt.title("ResNet50 Training and Validation Metrics")
plt.xlabel("Epoch")
plt.ylabel("Value")
plt.legend(); plt.show()
```
