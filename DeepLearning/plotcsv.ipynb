```python
import pandas as pd
from matplotlib.ticker import MultipleLocator
import matplotlib.pyplot as plt

log = pd.read_csv("training_log_resnet50.csv")
plt.ylim(0,1)
plt.xlim(log[log.phase=="train"].epoch.min(), log[log.phase=="train"].epoch.max())
plt.gca().xaxis.set_major_locator(MultipleLocator(100))
plt.gca().yaxis.set_major_locator(MultipleLocator(0.1))
ax = plt.gca()
for x in ax.xaxis.get_majorticklocs():
    ax.axvline(x, color='gray', linestyle='solid', linewidth=0.6, alpha=0.5)
for y in ax.yaxis.get_majorticklocs():
    ax.axhline(y, color='gray', linestyle='solid', linewidth=0.6, alpha=0.5)
plt.plot(log[log.phase=="train"].epoch, log[log.phase=="train"].loss, label="training loss")
plt.plot(log[log.phase=="val"].epoch, log[log.phase=="val"].loss, label="validation loss")
plt.plot(log[log.phase=="train"].epoch, log[log.phase=="train"].accuracy, label="training accuracy")
plt.plot(log[log.phase=="val"].epoch, log[log.phase=="val"].accuracy, label="validation accuracy")
plt.title("ResNet50 Training and Validation Metrics")
plt.xlabel("Epoch")
plt.ylabel("Value")
plt.legend(); plt.show()
plt.savefig("resnet50.png")

# find the epoch with the best validation accuracy and print the accuracy false positive rate and false negative rate at that epoch
best_epoch = log[log.phase=="val"].accuracy.idxmax()
best_accuracy = log.loc[best_epoch, "accuracy"]
best_fpr = log.loc[best_epoch, "false_positive_rate"]
best_fnr = log.loc[best_epoch, "false_negative_rate"]
print(f"Best Epoch: {log.loc[best_epoch, 'epoch']}")
print(f"Best Validation Accuracy: {best_accuracy*100:.2f}%")
print(f"False Positive Rate at Best Epoch: {best_fpr*100:.2f}%")
print(f"False Negative Rate at Best Epoch: {best_fnr*100:.2f}%")
```
